### Open original data

### Cut data (if we don't look at the whole cube)

```python
if not np.array_equal(ecc, np.array([[0,75], [0,75], [0,75]])):

    mask = ((data[:,0] >= ecc[0,0]) & (data[:,0] <= ecc[0,1]) & 
            (data[:,1] >= ecc[1,0]) & (data[:,1] <= ecc[1,1]) &
            (data[:,2] >= ecc[2,0]) & (data[:,2] <= ecc[2,1]))

    data = data[mask] - np.array([ecc[0,0], ecc[1,0], ecc[2,0]])
```

### Convert coord to cell index  -->  Count index occurances = Get counts --> Sum to get convert to grid counts

*By not nesting the functions below, we save ~1.5Gb RAM for the 256 frames.


Reasons for that:

1. *Intermediate Objects*: Each function call creates a new object as its return value.
2. *Stack Frames*: Python creates a stack frame for each function call. In nested calls, all three functions are on the call stack at the same time, each with its own local variables and context, increasing memory usage.
3. *Garbage Collection Delay*: Python's garbage collector may not immediately reclaim memory from intermediate objects because they are still referenced until the nested expression completes.

```python
data = [fct_coord_to_index(i, d_xyz, size) for i in data]
```

```python
data = fct_index_to_counts(data, size)
```

```python
data = fct_list_to_grid(data, size)
```
