```python
output_count = -1
for sz_indx, size in enumerate(ALL_sizes):



    
    print(); print(); print(); print("- - - - - - - - - - - - - - - - - - sz_indx", sz_indx); print()

    cores_no___1_NGP                   = ALL_cores_no___1_NGP[                  sz_indx]
    cores_no___2_Smoothing             = ALL_cores_no___2_Smoothing[            sz_indx]
    cores_no___3_Layer                 = ALL_cores_no___3_Layer[                sz_indx]
    cores_no___4_Origins               = ALL_cores_no___4_Origins[              sz_indx]
    cores_no___5_UOD_vals_and_lvls     = ALL_cores_no___5_UOD_vals_and_lvls[    sz_indx]
    cores_no___6_Levels_isolated_pairs = ALL_cores_no___6_Levels_isolated_pairs[sz_indx]
    cores_no___7_Finder                = ALL_cores_no___7_Finder[               sz_indx]

    
    file_path   = ALL_file_paths[ sz_indx]
    
    Zs          = ALL_Zs[         sz_indx]
    Zs_snapshot = ALL_Zs_snapshot[sz_indx]
    d_xyz       = ALL_d_xyzs[     sz_indx]
    eccs        = ALL_eccs[       sz_indx]
    
    nncs        = ALL_nncs[       sz_indx]
    
    R_cMpchs    = ALL_R_cMpchs[   sz_indx]
    R_cells     = ALL_R_cells[    sz_indx]
    
    lvls        = ALL_lvls[       sz_indx]
    cdf_logs    = ALL_cdf_logs[   sz_indx]
    
    run_Origins = ALL_run_Origins[sz_indx]
    uods        = ALL_uods[       sz_indx]
    run_levels  = ALL_run_levels[ sz_indx]
    
    run_MKs     = ALL_run_MKs[    sz_indx]
    save1percs  = ALL_save1percs[ sz_indx]




    
    if RUN_NGP:
        print("- - - - - - - - - - - - - - - - - - Nearest_Grid_Point")
        notebook_filenames = []; notebook_parameters = []
        for     i0, Z   in enumerate(Zs):
            for i1, nnc in enumerate(nncs[i0]):
                
                Z_snapshot = Zs_snapshot[i0]
                
                file_path_Z   = file_path  +"Z___"  +Z       +"/"
                file_path_NNC = file_path_Z+"NNC___"+str(nnc)+"/"; make_folder(file_path_NNC)
                
                if nnc == 0:
                    notebook_filenames.append("1___Nearest_Grid_Point.ipynb")
                    notebook_parameters.append({"base_path_snapshots": base_path_snapshots,
                                                "size":                size,
                                                "Z_snapshot":          Z_snapshot,
                                                "ecc":                 np.array2string(np.array(ecc), separator=","),
                                                "d_xyz":               d_xyz,
                                                "file_path_NNC":       file_path_NNC})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___1_NGP)




    
    if RUN_Smoothing:
        print("- - - - - - - - - - - - - - - - - - Smoothing")
        notebook_filenames = []; notebook_parameters = []
        for         i0, Z       in enumerate(Zs):
            for     i1, nnc     in enumerate(nncs[    i0]):
                for i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    
                    R_cell = R_cells[i0][i1][i2]
                    
                    file_path_Z   = file_path    +"Z___"  +Z           +"/"
                    file_path_NNC = file_path_Z  +"NNC___"+str(nnc)    +"/"
                    file_path_R   = file_path_NNC+"R___"  +str(R_cMpch)+"/"; make_folder(file_path_R)
                    
                    notebook_filenames.append("2___Smoothing.ipynb")
                    notebook_parameters.append({"R_cell":                  R_cell,
                                                "parameter_median_filter": parameter_median_filter,
                                                "parameter_grey_opening":  parameter_grey_opening,
                                                "parameter_grey_closing":  parameter_grey_closing,
                                                "nnc":                     nnc,
                                                "file_path_NNC":           file_path_NNC,
                                                "file_path_R":             file_path_R,
                                                "rho_delta":               rho_delta,
                                                "n_DM":                    n_DM,
                                                "m_DM":                    m_DM,
                                                "d_xyz":                   d_xyz})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___2_Smoothing)




    
    if RUN_Layer:
        print("- - - - - - - - - - - - - - - - - - Layer")
        notebook_filenames = []; notebook_parameters = []
        for                 i0, Z       in enumerate(Zs):
            for             i1, nnc     in enumerate(nncs[    i0]):
                for         i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    for     i3, lvl     in enumerate(lvls[    i0][i1][i2]):
                        for i4, cdf_log in enumerate(cdf_logs[i0][i1][i2][i3]):
                            
                            file_path_Z   = file_path    +"Z___"  +Z           +"/"
                            file_path_NNC = file_path_Z  +"NNC___"+str(nnc)    +"/"
                            file_path_R   = file_path_NNC+"R___"  +str(R_cMpch)+"/"
                            file_path_Lvl = file_path_R  +"Lvl___"+str(lvl)    +"/"+cdf_log+"/"; make_folder(file_path_Lvl)
                            
                            notebook_filenames.append("3___Layer.ipynb")
                            notebook_parameters.append({"lvl":           lvl,
                                                        "file_path_R":   file_path_R,
                                                        "file_path_Lvl": file_path_Lvl,
                                                        "cl":            cdf_log,
                                                        "rho_delta":     rho_delta})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___3_Layer)





    if RUN_Origins:
        print("- - - - - - - - - - - - - - - - - - Origins")
        notebook_filenames = []; notebook_parameters = []
        for                 i0, Z       in enumerate(Zs):
            for             i1, nnc     in enumerate(nncs[    i0]):
                for         i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    for     i3, lvl     in enumerate(lvls[    i0][i1][i2]):
                        for i4, cdf_log in enumerate(cdf_logs[i0][i1][i2][i3]):

                            run_Origin    = run_Origins[i0][i1][i2][i3][i4]
                            
                            file_path_Z   = file_path    +"Z___"  +Z           +"/"
                            file_path_NNC = file_path_Z  +"NNC___"+str(nnc)    +"/"
                            file_path_R   = file_path_NNC+"R___"  +str(R_cMpch)+"/"
                            file_path_Lvl = file_path_R  +"Lvl___"+str(lvl)    +"/"+cdf_log+"/"
                            
                            if run_Origin:
                                notebook_filenames.append("4___Origins.ipynb")
                                notebook_parameters.append({"lvl":           lvl,
                                                            "file_path_Lvl": file_path_Lvl})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___4_Origins)




        
    if RUN_UOD_vals_and_lvls:
        print("- - - - - - - - - - - - - - - - - - UOD_vals_and_lvls")
        notebook_filenames = []; notebook_parameters = []
        for                 i0, Z       in enumerate(Zs):
            for             i1, nnc     in enumerate(nncs[    i0]):
                for         i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    for     i3, lvl     in enumerate(lvls[    i0][i1][i2]):
                        for i4, cdf_log in enumerate(cdf_logs[i0][i1][i2][i3]):

                            uods_i = uods[i0][i1][i2][i3][i4]
                            
                            file_path_Z   = file_path    +"Z___"  +Z           +"/"
                            file_path_NNC = file_path_Z  +"NNC___"+str(nnc)    +"/"
                            file_path_R   = file_path_NNC+"R___"  +str(R_cMpch)+"/"
                            file_path_Lvl = file_path_R  +"Lvl___"+str(lvl)    +"/"+cdf_log+"/"
                            
                            if uods_i != [[]] and uods_i != []:
                                notebook_filenames.append("5___UOD_vals_and_lvls.ipynb")
                                notebook_parameters.append({"lvl":           lvl,
                                                            "uods":          np.array2string(np.array(uods_i), separator=","),
                                                            "file_path_R":   file_path_R,
                                                            "file_path_Lvl": file_path_Lvl,
                                                            "rho_delta":     rho_delta})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___5_UOD_vals_and_lvls)





    if RUN_Levels_isolated_pairs:
        print("- - - - - - - - - - - - - - - - - - Levels_isolated_pairs")
        notebook_filenames = []; notebook_parameters = []
        for                 i0, Z       in enumerate(Zs):
            for             i1, nnc     in enumerate(nncs[    i0]):
                for         i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    for     i3, lvl     in enumerate(lvls[    i0][i1][i2]):
                        for i4, cdf_log in enumerate(cdf_logs[i0][i1][i2][i3]):

                            run_level = run_levels[i0][i1][i2][i3][i4]
                            
                            file_path_Z   = file_path    +"Z___"  +Z           +"/"
                            file_path_NNC = file_path_Z  +"NNC___"+str(nnc)    +"/"
                            file_path_R   = file_path_NNC+"R___"  +str(R_cMpch)+"/"
                            file_path_Lvl = file_path_R  +"Lvl___"+str(lvl)    +"/"+cdf_log+"/"
                            
                            if run_level:
                                make_folder(file_path_Lvl+"dict_cells_pairs/")
                                make_folder(file_path_Lvl+"dict_cells_isolated/")
                                notebook_filenames.append("6___Levels_isolated_pairs.ipynb")
                                notebook_parameters.append({"lvl":           lvl,
                                                            "file_path_Lvl": file_path_Lvl})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___6_Levels_isolated_pairs)




    
    if RUN_Finder:
        print("- - - - - - - - - - - - - - - - - - Finder")
        make_folder("./progress")
        notebook_filenames = []; notebook_parameters = []
        for                             i0, Z       in enumerate(Zs):
            for                         i1, nnc     in enumerate(nncs[    i0]):
                for                     i2, R_cMpch in enumerate(R_cMpchs[i0][i1]):
                    for                 i3, lvl     in enumerate(lvls[    i0][i1][i2]):
                        for             i4, cdf_log in enumerate(cdf_logs[i0][i1][i2][i3]):

                            uods_i = uods[i0][i1][i2][i3][i4]
                            
                            if uods_i != [[]]:
                                for     i5, uod     in enumerate(uods_i):
                                    for i6, MK      in enumerate(run_MKs[ i0][i1][i2][i3][i4][i5]):
                                        
                                        ud, od = uod
                                        MK = "MK"+str(MK)
                                        save1perc = save1percs[i0][i1][i2][i3][i4][i5][i6]
                                        
                                        file_path_Z    = file_path     +"Z___"  +Z                  +"/"
                                        file_path_NNC  = file_path_Z   +"NNC___"+str(nnc)           +"/"
                                        file_path_R    = file_path_NNC +"R___"  +str(R_cMpch)       +"/"
                                        file_path_Lvl  = file_path_R   +"Lvl___"+str(lvl)           +"/"+cdf_log+"/"
                                        file_path_Duod = file_path_Lvl +"D___[" +str(ud)+"_"+str(od)+"]/"
                                        file_path_MK   = file_path_Duod+MK                          +"/"; make_folder(file_path_MK)
                                        
                                        output_count += 1
                                        if True:#not os.path.exists(file_path_MK+"fg___100.pk"):
                                            notebook_filenames.append("7___Finder.ipynb")
                                            notebook_parameters.append({"MK":                 MK,
                                                                        "size":               size,
                                                                        "ud":                 ud,
                                                                        "od":                 od,
                                                                        "times_we_tried_max": times_we_tried_max,
                                                                        "file_path_Lvl":      file_path_Lvl,
                                                                        "file_path_MK":       file_path_MK,
                                                                        "output_count":       output_count,
                                                                        "save1perc":          save1perc,
                                                                        "override_uod":       False,
                                                                        "ud_lvl"      :       -1,
                                                                        "od_lvl"      :       -1})
        run_the_notebooks(running, notebook_filenames, notebook_parameters, notebook_directory, output_directory, cores_no___7_Finder)
```
